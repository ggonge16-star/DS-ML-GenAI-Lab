# Creating IQR reference for outlier analysis

```python
def build_iqr_ref(df, cols):
    ref = {}
    for col in cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5*IQR
        upper = Q3 + 1.5*IQR

        ref[col] = {'lower': lower, 'upper': upper}
    return ref

iqr_ref = build_iqr_ref(df,nos_col)

```

## 4.What is the distribution of key air pollutants (PM2.5, PM10, NO₂, CO)?

```python
pollutants = ['PM2_MAX', 'PM10_MAX', 'NO2_MAX', 'CO_MAX']

for col in pollutants:

    # IQR reference 
    lower = iqr_ref[col]['lower']
    upper = iqr_ref[col]['upper']

    # 1️⃣ With outliers (original data)
    plt.figure(figsize=(4,3))
    df[col].hist()
    plt.title(col + " (with outliers)")
    plt.show()

    # 2️⃣ Without outliers (temporary filter,)
    temp = df[(df[col] >= lower) & (df[col] <= upper)]

    plt.figure(figsize=(4,3))
    temp[col].hist()
    plt.title(col + " (without outliers)")
    plt.show()
```

## 5 What is the overall distribution of environmental factors like temperature, humidity, sound, and light?

```python
env = ['TEMPRATURE_MAX','HUMIDITY','SOUND','LIGHT']
for i in env:
   # IQR reference 
    lower = iqr_ref[i]['lower']
    upper = iqr_ref[i]['upper']
    
    # 1️⃣ With outliers (real distribution)
    plt.figure(figsize=(4,3))
    plt.hist(df[i])
    plt.title(i + " (with outliers)")
    plt.show()

    # 2️⃣ Without outliers (clear shape)
    temp = df[(df[i] >= lower) & (df[i] <= upper)]
    plt.figure(figsize=(4,3))
    plt.hist(temp[i])
    plt.title(i + " (without outliers)")
    plt.show()
    
```

# 7.How does PM2.5 vary with temperature and humidity? 

```python
lower = iqr_ref['PM2_MAX']['lower']
upper = iqr_ref['PM2_MAX']['upper']

clean = df[(df['PM2_MAX'] >= lower) & (df['PM2_MAX'] <= upper)]
fig, ax = plt.subplots(2, 2, figsize=(7,5))

# WITH outliers
ax[0,0].scatter(df['TEMPRATURE_MAX'], df['PM2_MAX'], alpha=0.3)
ax[0,0].set_title("PM2.5 vs Temperature (with outliers)")

ax[0,1].scatter(df['HUMIDITY'], df['PM2_MAX'], alpha=0.3)
ax[0,1].set_title("PM2.5 vs Humidity (with outliers)")

# WITHOUT outliers (IQR reference)
ax[1,0].scatter(clean['TEMPRATURE_MAX'], clean['PM2_MAX'], alpha=0.3)
ax[1,0].set_title("PM2.5 vs Temperature (without outliers)")

ax[1,1].scatter(clean['HUMIDITY'], clean['PM2_MAX'], alpha=0.3)
ax[1,1].set_title("PM2.5 vs Humidity (without outliers)")

plt.tight_layout()
plt.show()
```
