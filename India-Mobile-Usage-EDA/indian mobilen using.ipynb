## numerical & categorical columns DIstributin of data

```python
numerical_columns = df.select_dtypes(include=['int32','float64']).columns.tolist()
catergorical_columns= df.select_dtypes(include=['object']).columns.tolist()
print(numerical_columns)
print('-'*50)
print(catergorical_columns)
```

```python
# Find outlier with IQR
```

```python
for i in numerical_columns:
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR =Q3 - Q1
    lower = Q1-1.5*IQR
    upper = Q3 +1.5*IQR
    cline = df[(df[i] <= lower ) | (df[i] >= upper)].shape[0]
    print(i,' ',lower,' ',upper, '   ', cline)
```

```python
#refrence
def iqr(df,cols):
    ref={}
    for g in cols:
        Q1 = df[g].quantile(0.25)
        Q3 = df[g].quantile(0.75)
        IQR = Q3 - Q1
        lower  = Q1 - 1.5 * IQR
        upper  = Q3 + 1.5 * IQR
        ref[g] = {'lower':lower , 'upper':upper}
    return ref
    
    
iqr_ref = iqr(df,numerical_columns)      
```
